35. Examination performance analysis: Generate a reproducible 12 x 4 NumPy array of integer test marks from
20 to 100, inclusive, using a fixed seed. Calculate each student's average and each test's mean and standard
deviation. Count students passing all four tests (at least 40 in each). Identify students with averages of at least 75
and the test with greatest variation. State your student numbering convention.

```python
import numpy as np

# Set fixed seed for reproducibility
np.random.seed(42)

# Generate 12 x 4 array of integer test marks from 20 to 100, inclusive
marks = np.random.randint(20, 101, size=(12, 4))

# 1. Calculate each student's average (mean across columns, axis=1)
student_averages = np.mean(marks, axis=1)

# 2. Calculate each test's mean and standard deviation (across rows, axis=0)
test_means = np.mean(marks, axis=0)
test_stds = np.std(marks, axis=0)

# 3. Count students passing all four tests (at least 40 in each)
passing_all = np.all(marks >= 40, axis=1)
num_passing_all = np.sum(passing_all)

# 4. Identify students with averages of at least 75
high_avg_students = np.where(student_averages >= 75)[0]

# 5. Identify the test with the greatest variation (highest standard deviation)
greatest_var_test = np.argmax(test_stds)

print("Generated Marks Matrix (12 students x 4 tests):\n", marks)
print("\nStudent Averages:", np.round(student_averages, 2))
print("Test Means:", np.round(test_means, 2))
print("Test Standard Deviations:", np.round(test_stds, 2))
print("Number of students passing all tests:", num_passing_all)
print("Students with average >= 75 (Student IDs):", high_avg_students + 1)
print("Test with greatest variation: Test", greatest_var_test + 1)
```

37. Sales and revenue analysis: Create a 6 x 4 matrix of quantities sold over six days for four products. Use unit
prices [150, 240, 90, 320]. Calculate daily revenue, total revenue per product, the highest-revenue day, and the
best-selling product by quantity. Reduce prices by 10% for products whose total quantity exceeds 100, then
recalculate total revenue without modifying original prices.

```python
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# 1. Create a 6 x 4 matrix of quantities sold (6 days rows x 4 products columns)
quantities = np.random.randint(10, 36, size=(6, 4))

# Unit prices for the four products
prices = np.array([150, 240, 90, 320])

# 2. Calculate daily revenue (summing across products per day using broadcasting/matrix multiplication)
daily_revenue = np.sum(quantities * prices, axis=1)

# 3. Calculate total revenue per product
total_quantity_per_product = np.sum(quantities, axis=0)
total_revenue_per_product = total_quantity_per_product * prices

# 4. Identify the highest-revenue day
highest_day_idx = np.argmax(daily_revenue)
highest_day = highest_day_idx + 1

# 5. Identify the best-selling product by quantity
best_product_idx = np.argmax(total_quantity_per_product)
best_product = best_product_idx + 1

# 6. Reduce prices by 10% for products with total quantity > 100 (without modifying original prices)
exceeds_100 = total_quantity_per_product > 100
adjusted_prices = np.where(exceeds_100, prices * 0.9, prices)

# Recalculate total revenue with adjusted prices
adjusted_revenue_per_product = total_quantity_per_product * adjusted_prices
overall_original_revenue = np.sum(total_revenue_per_product)
overall_adjusted_revenue = np.sum(adjusted_revenue_per_product)

# Print Results
print("Quantities Matrix (6 Days x 4 Products):\n", quantities)
print("\nDaily Revenue:", daily_revenue)
print("Total Quantity per Product:", total_quantity_per_product)
print("Total Revenue per Product:", total_revenue_per_product)
print(f"Highest-Revenue Day: Day {highest_day} ({daily_revenue[highest_day_idx]})")
print(f"Best-Selling Product by Quantity: Product {best_product} ({total_quantity_per_product[best_product_idx]} units)")
print("\n--- Discount Analysis (Total Quantity > 100) ---")
print("Original Prices:", prices)
print("Adjusted Prices (10% off for >100 units):", adjusted_prices)
print("Overall Revenue (Original):", overall_original_revenue)
print("Overall Revenue (Adjusted):", overall_adjusted_revenue)
```

38. Random data cleaning and statistics: Generate 100 normal observations with mean 50 and standard
deviation 12 using a fixed seed. Count values outside 30-70 and replace them with the nearest boundary using
np.where(). Compare original and cleaned mean, variance, standard deviation, minimum, and maximum. Explain
the statistical effects of boundary replacement.

```python
import numpy as np
import pandas as pd

# Set fixed seed for reproducibility
np.random.seed(42)

# 1. Generate 100 normal observations with mean 50 and standard deviation 12
data = np.random.normal(loc=50, scale=12, size=100)

# 2. Count values outside the boundary 30 to 70
outside_mask = (data < 30) | (data > 70)
count_outside = np.sum(outside_mask)

# 3. Replace values outside 30-70 with the nearest boundary using np.where()
cleaned_data = np.where(data < 30, 30, np.where(data > 70, 70, data))

# 4. Compare statistical metrics
comparison = pd.DataFrame({
    'Metric': ['Mean', 'Variance', 'Standard Deviation', 'Minimum', 'Maximum'],
    'Original': [
        np.mean(data),
        np.var(data),
        np.std(data),
        np.min(data),
        np.max(data)
    ],
    'Cleaned': [
        np.mean(cleaned_data),
        np.var(cleaned_data),
        np.std(cleaned_data),
        np.min(cleaned_data),
        np.max(cleaned_data)
    ]
})

print(f"Number of values outside 30-70: {count_outside}")
print("\n", comparison.to_string(index=False))
```

39. Multi-branch financial reconciliation: Combine the three earnings Series by ordinary addition. Explain missing
values, identify quarters common to all branches, and calculate their combined earnings. Construct a separate
result treating absent branch records as zero and explain how its interpretation differs.
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})

```python
import pandas as pd

scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']

# 1. Create Series preserving required index order
s = pd.Series(scores, index=required)

# 2. Identify missing student records (NaN values)
missing_students = s[s.isna()].index.tolist()

# 3. Filter valid scores of at least 70
filtered_s = s[s >= 70]

# 4. Convert to a dictionary and reconstruct a new Series
d = filtered_s.to_dict()
reconstructed_s = pd.Series(d)

# 5. Verify that labelled values match
values_match = filtered_s.equals(reconstructed_s)

print("Initial Series:\n", s)
print("\nMissing student records:", missing_students)
print("\nFiltered Series (>= 70):\n", filtered_s)
print("\nReconstructed Series:\n", reconstructed_s)
print("\nLabelled values match:", values_match)
```
