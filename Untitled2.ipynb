```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load the image
img = Image.open("Screenshot.png").convert("L")

# Make it small
#img = img.resize((60, 80))

# Convert image → NumPy array
arr = np.array(img)*255

print(arr.shape)
print(arr)

# Convert NumPy array → image
plt.imshow(arr, cmap="gray")
plt.axis("off")
plt.show()
```
